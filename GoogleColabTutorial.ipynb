```
# Defining functions in Collab
inputFib = int(input("Enter a number for Fibonacci calculation: "))

def fib(inputFib):
  #Printing a fibonacci series...
  a, b = 0, 1

  while a < inputFib:
    print(a, end=' ')
    a, b = b, a+b

  print()

#Calling fib function
fib(inputFib)
```

```
# Write a Python program to find all the permutations of a given string.
from itertools import permutations

def string_permutations(s):
  return [''.join(p) for p in permutations(s)]

#Test case
print(string_permutations("plane"))
```

```
# Write a Python function is_palindrome

def is_palindrome(s):
  cleaned = ''.join(c.lower() for c in s if c.isalnum())
  return cleaned == cleaned[::-1]

# Test cases
print(is_palindrome("A mean a plan a canal Panama"))
print(is_palindrome("Python"))
print(is_palindrome("Level"))
```

```
# Write a function fizz_buzz that prints the numbers from 1 to a given number n. But:
# - For multiples of 3, print 'Fizz' instead of the number.
# - For multiples of 5, print 'Buzz' instead of the number.
# - For multiples of both 3 & 5, print 'FizzBuzz'.


def fizz_buzz(n):
  for i in range(1, n):
    if (i%3 == 0):
      print(i)
      print("Fizz")

    elif (i%5 == 0):
      print(i)
      print("Buzz")

    elif (i%3 == 0) and (i%5 == 0):
      print(i)
      print("FizzBuzz")

    else:
      print(i)


n = int(input('Enter your number: '))
fizz_buzz(n)
```

# Numpy fundamentals

Read numpy_fundamentals.pdf and create code cells below to experiment with Numpy.

Numpy's main object is the multidimensional array (ndarray).

** Is a table of elements (all of the same data type), indexed by a tuple of positive integers.

```
import numpy as np

arr = np.array(range(1, 10))
print('arr\t', arr)

print('Rank of the matrix: \t', arr.ndim) #1 means 1D array
print('Shape of the matrix: \t', arr.shape) #(9,) means 9 rows, 0 columns

print('Size of the matrix: \t', arr.size) #the total number of elements in the array (aka the product of the dimensions)
print('Type of elements in the array: \t', arr.dtype) #int64 means integer datatype
```

```
#Creating a 1D array
oneD = np.array([1, 2, 3, 4])
print('Rank of the matrix: ', oneD.ndim)
print('Shape of the matrix: ', oneD.shape)
oneD
print('')

twoD = np.array([[1, 2], [3, 4], [5, 6]])
print('Rank of the matrix: ', twoD.ndim)
print('Shape of the matrix: ', twoD.shape)
twoD
```

```
# When you need to create a array, but only know its dimensions. Initialise to zeros, ones or empty.

zero_np = np.zeros((2, 3)) #all elements are explicitly initialised to 0
print("Zero_np")
print(zero_np)

one_np = np.ones((2, 3)) #all elements are explicitly initialised to 1
print("One_np")
print(one_np)

empty_np = np.empty((2, 3)) #all elements will have random values
print("Empty_np")
print(empty_np)
```

```
# arrange() returns a sequence of numbers (like range() )
np.arange(1, 3, 0.2)


# Alternatively, use linspace (for creating floating numbers sequence)
np.linspace(1, 3, 7)
```

```
# Reshaping the ndarray

arr = np.arange(1, 10)
print("Old shape: ", arr)
print("Shape: ", arr.shape)
```

```
newShape = arr.reshape((3, 3)) #Returns a new ndarray without modifying the original one
print("New shape: \n", newShape)
print('')

arr
```
